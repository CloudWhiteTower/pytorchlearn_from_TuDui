```python
# 用来演示最大池化层的使用

import torch
from torch import nn
import torchvision as tv
from torch.utils.tensorboard import SummaryWriter
from torch.utils.data import DataLoader
```

```python
x = torch.tensor([[0., 1., 2., 3., 4.],
                   [3., 4., 5., 6., 7.],
                   [6., 7., 8., 9., 10.],
                   [9., 10., 11., 12., 13.],
                   [12., 13., 14., 15., 16.]])

x = torch.reshape(x, (-1, 1, 5, 5))

dataset = tv.datasets.CIFAR10(root="./dataset", train=False, transform=tv.transforms.ToTensor(), download=True)
dataloader = DataLoader(dataset, batch_size=64)

# 定义一个最大池化层，kernel_size=3，ceil_mode=True
class tudui(nn.Module):
    def __init__(self):
        super().__init__()
        self.maxpool = nn.MaxPool2d(kernel_size=3, ceil_mode=True)

    def forward(self, x):
        y = self.maxpool(x)
        return y
    

obj1 = tudui()

writer = SummaryWriter("logs/experiment5_maxpool")
step = 0


for data in dataloader:
    img, label = data
    writer.add_images("ip_img", img, step)
    out = obj1(img)
    writer.add_images("op_img", out, step)

    step += 1

writer.close()
```
