```python
# 如何使用torchvision 给我们提供的高效数据集进行训练

import torchvision as tv
from torch.utils.tensorboard import SummaryWriter

data_trasnform = tv.transforms.Compose([
    tv.transforms.ToTensor()])

trans_set = tv.datasets.CIFAR10(
    root="./dataset",
    transform=data_trasnform,
    train=True,
    download=True,)

test_set = tv.datasets.CIFAR10(
    root="./dataset",
    transform=data_trasnform,
    train=False,
    download=True,)

img, target = trans_set[0]
print(img, target)
print(test_set.classes[target])
```

```python
write  = SummaryWriter("./logs/experiment2")
for i in range(10):
    img, target = trans_set[i]
    write.add_image("CIFAR10/{}".format(test_set.classes[target]), img, i)
write.close()
```
