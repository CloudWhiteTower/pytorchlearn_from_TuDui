```python
# 用于卷积神经网络的示例代码
import torch
from torch.nn import functional as F

```

```python

x = torch.tensor([[1, 2, 0, 3, 1],
                  [0, 1, 2, 3, 1],
                  [1, 2, 1, 0, 0],
                  [5, 2, 3, 1, 1],
                  [2, 1, 0, 1, 1]])

kernel = torch.tensor([[1,0,1],
                       [0,1,0],
                       [1,0,1]])

# 用reshape将kernel调整为卷积层所需的形状，为(batch, 通道数, 高度, 宽度)，要不然会报错

x = torch.reshape(x, (1,1,5,5))
kernel = torch.reshape(kernel, (1,1,3,3))

# 使用F.conv2d进行卷积操作(不同于nn.Conv2d模块)
out = F.conv2d(x, kernel, stride=1, padding=0)
print(out)  # 输出卷积结果
# 如果需要保持输入和输出的尺寸相同，可以使用padding=1
out1 = F.conv2d(x, kernel, stride=1, padding=1)
print(out1)  # 输出卷积结果
```
