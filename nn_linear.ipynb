```python
"""
学了这么多，基本都会用了。小土堆的视频里提到了批量归一层，Transformer，RNN等等，
这些层都可以看着文档学一学了，使用方法都差不多
下面是线性层（全连接层）的讲解和展平层的讲解。展平层flatten可以吧多维的输入展平为一维的输入，线性层就是全连接层了，输入输出都是一维的。
"""
import torch
import torchvision as tv
from torch import nn
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
```

```python
dataser = tv.datasets.CIFAR10(root="./dataset", train=True, transform=tv.transforms.ToTensor(), download=True)
dataloader = DataLoader(dataser, batch_size=64, drop_last=True)

class tudui(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear1 = nn.Linear(196608,10)

    def forward(self, x):
        y  = self.linear1(x)
        return y

model = tudui()

for data in dataloader:
    img, target = data
    op = torch.flatten(img)
    op = model(op)
    print(op.shape)
```
