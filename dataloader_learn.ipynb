```python
import torchvision as tv
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
```

```python
# 准备的测试数据集
test_data = tv.datasets.CIFAR10(
    root="./dataset",
    train=False,
    transform=tv.transforms.ToTensor(),
    download=True,
)
# 如何使用 dataloader
test_dataloader = DataLoader(dataset=test_data, batch_size=64, shuffle=True, num_workers=0, drop_last=False)

writer = SummaryWriter("logs/experiment3")
global_step = 0
for data in test_dataloader:
    imgs, targets = data
    # print(imgs.shape)
    # print(targets)
    writer.add_images("test_imgs", imgs, global_step=global_step, dataformats="NCHW")
    global_step += 1

writer.close()
```
