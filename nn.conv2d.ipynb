```python
# 下面是 nn.Module 的使用示例，需要输入（in_channels, out_channels, kernel_size, stride, padding，等等参数）
# 其中，out_channels 表示卷积核的数量，也就是输出的通道数。会生成多个卷积核以生成多个特征图（也就是多个输出通道）
import torch
import torchvision as tv
from torch import nn
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
```

```python
from torch import conv2d


dataset = tv.datasets.CIFAR10(root='./dataset', train=False, download=True, transform=tv.transforms.ToTensor())
dataloader = DataLoader(dataset, batch_size=32)

# 定义一个简单的卷积神经网络
class tudui(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=6, kernel_size=3, stride=1, padding=0)

    def forward(self, x):
        x = self.conv1(x)
        return x

obj1 = tudui()

writer = SummaryWriter(log_dir="./logs/experiment4")
step = 0

for data in dataloader:
    imgs, targets = data
    outputs = obj1(imgs)
    print("输入尺寸:", imgs.shape)
    print("输出尺寸:", outputs.shape)

    # 输入大小: torch.Size([32, 3, 32, 32])
    writer.add_images("input", imgs, step)

    # 输出大小: torch.Size([32, 6, 30, 30])
    # 注意 tensorboard 只能显示单通道或三通道的图片，所以这里我们需要使用 reshape 
    #reshape如果不知道input的具体shape，可以使用-1来自动计算
    outputs = torch.reshape(outputs, (-1,3,30,30))
    writer.add_images("output", outputs, step)
    step += 1
writer.close()
```
