```python
from  torchvision import  transforms
from PIL import Image
from torch.utils.tensorboard import SummaryWriter

writer = SummaryWriter("logs/experiment1", )

# 如何使用totensor将图片转换为tensor
img1 = Image.open("hymenoptera_data/train/ants/5650366_e22b7e1065.jpg")
img2 = Image.open("hymenoptera_data/train/ants/6240329_72c01e663e.jpg")
tensor_trans = transforms.ToTensor()
tensor_img1 = tensor_trans(img1)
tensor_img2 = tensor_trans(img2)
writer.add_image("tensor_img2", tensor_img2, 1)
writer.close()

#Normalize的使用 归一化，可以达到很诡异的效果，让图片纵情变化
trans_norm = transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])
img1_norm = trans_norm(tensor_img1)
writer.add_image("img1_norm", img1_norm, 2)
writer.close()

# Resize的使用 改变图片的尺寸
img_trans_resize = transforms.Resize((512, 512))
img1_resize = img_trans_resize(tensor_img1)
writer.add_image("img1_resize", img1_resize, 3)
writer.close()

# Compose的使用 将多种操作组合在一起。他需要一个列表作为参数，使用方法是a = transforms.Compose([操作1，操作2，...])
trans_rsize_2 = transforms.Resize((256, 256))
trans_to_tensor2 = transforms.ToTensor()
trans_compose = transforms.Compose([trans_rsize_2, trans_to_tensor2])
img2_compose = trans_compose(img2)
writer.add_image("img2_compose", img2_compose, 4)

# RandomCrop的使用 随机裁剪图片
trans_randomcrop = transforms.RandomCrop(216)
for i in range(6):
    img2_randomcrop = trans_randomcrop(tensor_img2)
    writer.add_image("img2_randomcrop", img2_randomcrop, i)
writer.close()
```
