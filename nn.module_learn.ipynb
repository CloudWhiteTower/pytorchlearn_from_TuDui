```python
import torch
from torch import nn
```

```python
# 定义一个简单的神经网络模块
class tudui(nn.Module):
    def __init__(self):
        super().__init__()
    
    def forward(self, x):
        y = x + 1
        return y

# 创建模块的实例
cloud = tudui()
x = torch.tensor(1.0)

"""
前向传播：推荐直接使用 cloud(x)，而不是 cloud.forward(x)
因为前者会自动处理一些内部细节，如 hooks 和 pre/post processing。
"""

y = cloud(x)
print(y)
```
