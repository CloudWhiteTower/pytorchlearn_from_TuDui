```python
# 用于学习loss函数的使用
import torch
from torch import nn
```

```python
ip = torch.tensor([0.1, 0.2, 0.3])
op = torch.tensor([0, 0, 0.3])

ip = ip.reshape(-1,1,1,3)
op = op.reshape(-1,1,1,3)

# L1loss是绝对值损失函数，MSELoss是均方误差损失函数
loss_fn = nn.L1Loss()
loss = loss_fn(ip, op)

loss_mse  = nn.MSELoss()
loss_mse_value = loss_mse(ip, op)
print(loss_mse_value)
```

```python
# 交叉熵损失函数，适用于分类问题
x = torch.tensor([0.1, 0.2, 0.3])
y = torch.tensor([1])

x = x.reshape(1,3)
loss_cross = nn.CrossEntropyLoss()
reslt_cross = loss_cross(x, y)
print(reslt_cross)
```
