```python
# 用于学习非线性激活，以relu 作为例子。
import torch
import torchvision as tv
from torch import nn
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
```

```python
# relu可以将负数变为0，正数不变。
x = torch.tensor([[-1,1],
                  [-0.5,0.5],])
x = torch.reshape(x, (-1, 1, 2, 2))

dataset = tv.datasets.CIFAR10(root="./dataset", train=False, transform=tv.transforms.ToTensor(), download=True)
dataloader = DataLoader(dataset, batch_size=64)
writer = SummaryWriter("logs/experiment6_relu_sigmoid")

class tudui(nn.Module):
    def __init__(self):
        super().__init__()
        # 其中relu函数有参数inplace，默认为False（不改变原变量，需要一个新变量接受），如果设置为True，则会改变原变量。
        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        y = self.sigmoid(x)
        return y

tudui = tudui()
step = 0

for img, label in dataloader:
    ip = img
    writer.add_images("input", ip, global_step=step)
    op = tudui(ip)
    writer.add_images("output", op, global_step=step)
    step += 1

writer.close()
```
